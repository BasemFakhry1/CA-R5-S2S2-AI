### imports

```python
import numpy as np
import matplotlib.pyplot as plt
```

### Linear Regression Class

```python
class LinearRegressionGD:

    def __init__(self, learning_rate=0.001, n_iters=100):
        self.learning_rate = learning_rate
        self.n_iters = n_iters
        self.theta_0 = 0
        self.theta_1 = 0
        self.sse_history = []

    def fit(self, X, y):

        n = len(X)

        for i in range(self.n_iters):

            y_pred = self.theta_0 + self.theta_1 * X

            error = y_pred - y

            sse = np.sum(error ** 2)
            self.sse_history.append(sse)

            gradient_0 = (2 / n) * np.sum(error)
            gradient_1 = (2 / n) * np.sum(error * X)

            self.theta_0 -= self.learning_rate * gradient_0
            self.theta_1 -= self.learning_rate * gradient_1

    def predict(self, X):

        return self.theta_0 + self.theta_1 * X

    def plot_training(self, X, y):

        plt.figure(figsize=(8, 5))
        plt.plot(self.sse_history)

        plt.xlabel("Iterations")
        plt.ylabel("SSE")
        plt.title("SSE over Iterations")

        plt.show()

        plt.figure(figsize=(8, 5))

        y_pred = self.predict(X)

        plt.scatter(X, y, label="Actual Data")
        plt.plot(X, y_pred, label="Regression Line")

        plt.xlabel("House Size (m²)")
        plt.ylabel("House Price (Thousands)")
        plt.title("Linear Regression")

        plt.legend()
        plt.show()
```

### Data

```python
X = np.array([50, 60, 70, 80, 90], dtype=float)

y = np.array([150, 180, 210, 240, 270], dtype=float)
```

### model

```python
model = LinearRegressionGD(
    learning_rate=0.001,
    n_iters=100
)
```

### Training

```python
model.fit(X, y)
```

### Parameters

```python
print("Theta 0:", model.theta_0)
print("Theta 1:", model.theta_1)
```

### Prediction

```python
prediction = model.predict(70)

print("Predicted price for 70 m²:", prediction)
```

### Visualization

```python
model.plot_training(X, y)
```
