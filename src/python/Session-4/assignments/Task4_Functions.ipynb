### 1. Write a function that inputs a number and prints the multiplication table of that number


```python
def multiplication_table(num):
    for x in range (1, 11):
        print(f"{num} * {x} = {num*x}")
multiplication_table(9)             
```

### 2. Write a program to print twin primes less than 1000. If two consecutive odd numbers are both prime then they are known as twin primes

```python
def twin_prime():

    for num in range(3, 1000, 2):

        if num > 1 and num + 2 > 1:

            prime1 = True
            prime2 = True

            for i in range(2, num):
                if num % i == 0:
                    prime1 = False
                    break

            for i in range(2, num + 2):
                if (num + 2) % i == 0:
                    prime2 = False
                    break

            if prime1 and prime2:
                print(num, num + 2)


twin_prime()
```

### 3. Write a program to find out the prime factors of a number. Example: prime factors of 56 - 2, 2, 2, 7

```python
def prime_factors(num):

    for i in range(2, num + 1):

        while num % i == 0:
            print(i)
            num = num // i


number = int(input("Enter a number: "))

prime_factors(number)
```

### 4. Write a function that converts a decimal number to binary number

```python
def decToBin(num):

    binary = ""

    while num != 0:

        if num % 2 == 0:
            binary = "0" + binary
        else:
            binary = "1" + binary

        num = num // 2

    print(binary)

decToBin(11)
```

### 5. A number is called perfect if the sum of proper divisors of that number is equal to the number. For example 28 is perfect number, since 1+2+4+7+14=28. Write a program to print all the perfect numbers in a given range

```python
def perfectNums(start, end):

    for num in range(start, end + 1):

        if num == 0:
            print(0)
            continue

        total = 0

        for i in range(1, num):
            if num % i == 0:
                total += i

        if total == num:
            print(num)


perfectNums(0, 100)
```
