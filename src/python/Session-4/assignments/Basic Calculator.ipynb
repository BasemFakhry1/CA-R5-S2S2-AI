# Basic Calculator Task

### - Input

```python
def take_inputs():
    print("Welcome to the Simple Calculator!")
    print("Select an operation:")
    print("1. Addition (+)")
    print("2. Subtraction (-)")
    print("3. Multiplication (*)")
    print("4. Division (/)")
    choice = input("Enter your choice (1/2/3/4) or 'exit': ")
    return choice

def get_numbers():
    num_1 = float(input("Enter first number: "))
    num_2 = float(input("Enter second number: "))
    return num_1, num_2
```

### - Addition

```python
def addition(num_1, num_2):
    return num_1 + num_2
```

### - Subtraction

```python
def subtraction(num_1, num_2):
    return num_1 - num_2
```

### - Multiplication

```python
def multiplication(num_1, num_2):
    return num_1 * num_2
```

### - Division


```python
def division(num_1, num_2):
    if num_2 == 0:
        print("Error: Cannot divide by zero!")
        return None
    return num_1 / num_2
```

### - Main

```python

while True:
    choice = take_inputs()
    
    if choice == 'exit':
        print("Thank you!")
        break
    
    if choice == '1':
        num_1, num_2 = get_numbers()
        result = addition(num_1, num_2)
        print(f"Result: {num_1} + {num_2} = {result}\n")
    
    elif choice == '2':
        num_1, num_2 = get_numbers()
        result = subtraction(num_1, num_2)
        print(f"Result: {num_1} - {num_2} = {result}\n")
    
    elif choice == '3':
        num_1, num_2 = get_numbers()
        result = multiplication(num_1, num_2)
        print(f"Result: {num_1} * {num_2} = {result}\n")
    
    elif choice == '4':
        num_1, num_2 = get_numbers()
        result = division(num_1, num_2)
        if result is not None:
            print(f"Result: {num_1} / {num_2} = {result}\n")
```
